```python
import matplotlib.pyplot as plt
import pandas as pd
from pathlib import Path
from ollama_color_detection import colorDetection
from dataframe_creation import DataFrameCreator
from compute_statistics import compute_f1_and_hallucination
import time
import json

pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', None)
```

```python
prompt = (
    "You will be given a image featuring athletes."
    "From each image you should return the unique jersey colors that the athletes are wearing."
    "Return the colors hex code for all jersey shirts as well as the written color."
    "Ignore the shorts color and any undershirts, just include the shirt color that generally includes a number on it."
    "Don't return the same combinations of colors more than once for each response."
    
    "IMPORTANT INSTRUCTIONS:"
    "1. Only detect jersey colors that are FULLY VISIBLE in the image.\n"
    "2. Detect ALL of the jersey colors that are FULLY VISIBLE.\n"
    "3. Only detect jersey colors for ATHLETES in the image, ignore all others.\n"
    "4. Get the most specific color for the hex code that you can.\n"

    "RESPONSE FORMAT: Return a valid json for each unique jersey found in the image with the following format. "
    
    "hex_code_color: The color in hex of each element of the jersey"
    "color: The written out color\n"
    
    "Put each jersey (Can have multiple colors) into one element of the json."
    "Now analyze the image and return the json."
    
    "An example of this reponse would be:"
    "jerseys: { hex_code_color: [#0096FF, #FF0000], color: [Blue, Red]},\n"
    "{ hex_code_color: [#FFFFF, #FFA500], color: [White, Orange]}."
    "This is two different jerseys."
    
    "It should not be like:"
    "jerseys: { hex_code_color: [#0096FF, #0096FF], color: [Blue, Blue]}"
)
```

```python
image_folder = Path("images/")
image_files = [str(f) for f in image_folder.iterdir() if f.is_file()]
```

```python
qwen_df_creator = DataFrameCreator()
gemma_model = colorDetection(model_name="gemma3:12b")

start_time = time.time()

for img_path in image_files:
    print(img_path)
    gemma_output = gemma_model.extract_color_info(
        image_path=img_path,
        prompt=prompt
    )
    print(gemma_output)
```
